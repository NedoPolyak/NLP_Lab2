```python
import gensim
import re
from google.colab import drive
drive.mount('/content/drive')

word2vec = gensim.models.KeyedVectors.load_word2vec_format(
    "/content/drive/MyDrive/NLP/cbow.txt", binary=False
)
```

```python
TARGET = ("серия_NOUN", "пилот_NOUN")

pos = ["серия_NOUN"]
neg = []

pat = re.compile(r"^(.*)_NOUN$")

raw = word2vec.most_similar(positive=pos, negative=neg, topn=200)

nouns = []
for word, score in raw:
    m = pat.match(word)
    if m is not None:
        nouns.append((m.group(1), score))
    if len(nouns) == 10:
        break

print("10 существительных:")
for i, (w, s) in enumerate(nouns, 1):
    print(f"{i:2d}. {w:20s} {s:.4f}")
```
